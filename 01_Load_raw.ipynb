01_load_raw: პირველადი მონაცემების ჩატვირთვა getdata.raw სქემაში.

წყარო: The Himalayan Database (1905-2019), ნეპალის ჰიმალაი https://www.himalayandatabase.com/

ფაილები: /Volumes/getdata/raw/source_files/

ამ ეტაპზე მონაცემი ინახება უცვლელად, არანაირი ფილტრი,გაწმენდა ან გარდაქმნა არ ხდება. SELECT * EXCEPT (_rescued_data) გამოიყენება, რადგან read_files() ავტომატურად ამატებს ამ ტექნიკურ სვეტს, რომელიც წყაროს ფაილებში არ არსებობს.



```python
%sql
CREATE OR REPLACE TABLE getdata.raw.raw_peaks AS
SELECT * EXCEPT (_rescued_data)
FROM read_files(
    '/Volumes/getdata/raw/source_files/peaks.csv',
    format => 'csv',
    header => true,
    inferSchema => true
);
```

```python
%sql
CREATE OR REPLACE TABLE getdata.raw.raw_expeditions AS
SELECT * EXCEPT (_rescued_data)
FROM read_files(
    '/Volumes/getdata/raw/source_files/expeditions.csv',
    format  => 'csv',
    header  => true,
    inferSchema => true
);
```

```python
%sql
CREATE OR REPLACE TABLE getdata.raw.raw_members AS
SELECT * EXCEPT (_rescued_data)
FROM read_files(
    '/Volumes/getdata/raw/source_files/members.csv',
    format  => 'csv',
    header  => true,
    inferSchema => true
);
```

```python
%sql
SELECT *
FROM getdata.raw.raw_peaks;
```

```python
%sql
SELECT *
FROM getdata.raw.raw_expeditions;
```

```python
%sql
SELECT *
FROM getdata.raw.raw_members;
```

```python
%sql
SELECT 'raw_peaks' AS table_name, COUNT(*) AS row_count FROM getdata.raw.raw_peaks
UNION ALL
SELECT 'raw_expeditions', COUNT(*) FROM getdata.raw.raw_expeditions
UNION ALL
SELECT 'raw_members', COUNT(*) FROM getdata.raw.raw_members;
```
