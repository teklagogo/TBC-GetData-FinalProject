01_load_raw: პირველადი მონაცემების ჩატვირთვა getdata.raw სქემაში.


წყარო 1 : The Himalayan Database (1905-2024) ელიზაბეთ ჰოულის არქივი, https://www.himalayandatabase.com/ ჩამოწერილია Kaggle-დან (CC0), 00_download_source_data.py-ით

ფაილები: hdb_peaks.csv, hdb_exped.csv, hdb_members.csv

წყარო 2 : Wikipedia — Eight-thousander (CC BY-SA 4.0) + Wikidata (ბიოგრაფიული ფაქტები) + MediaWiki (ფოტოები)  ჩამოწერილია 00b_fetch_wikipedia_sources.py-ით

 ფაილები: peak_stats_8000ers.csv, peak_first_ascents.csv, summiters_14x8000.csv

ფაილები: /Volumes/getdata/raw/source_files/

ამ ეტაპზე მონაცემი ინახება უცვლელად, არანაირი ფილტრი,გაწმენდა ან გარდაქმნა არ ხდება. SELECT * EXCEPT (_rescued_data) გამოიყენება, რადგან read_files() ავტომატურად ამატებს ამ ტექნიკურ სვეტს, რომელიც წყაროს ფაილებში არ არსებობს.



```python
%sql
CREATE SCHEMA IF NOT EXISTS getdata.raw;
```

```python
%sql
CREATE OR REPLACE TABLE getdata.raw.raw_peaks AS
SELECT * EXCEPT (_rescued_data)
FROM read_files(
    '/Volumes/getdata/raw/source_files/hdb_peaks.csv',
    format => 'csv',
    header => true,
    inferSchema => true,
    multiLine => true,
    escape => '"'
);
```
