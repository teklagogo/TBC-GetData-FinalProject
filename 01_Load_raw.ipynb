01_load_raw: პირველადი მონაცემების ჩატვირთვა getdata.raw სქემაში.

წყარო: The Himalayan Database (1905-2019), ნეპალის ჰიმალაი https://www.himalayandatabase.com/

ფაილები: /Volumes/getdata/raw/source_files/

ამ ეტაპზე მონაცემი ინახება უცვლელად, არანაირი ფილტრი,გაწმენდა ან გარდაქმნა არ ხდება. SELECT * EXCEPT (_rescued_data) გამოიყენება, რადგან read_files() ავტომატურად ამატებს ამ ტექნიკურ სვეტს, რომელიც წყაროს ფაილებში არ არსებობს.



```python
%sql
CREATE OR REPLACE TABLE getdata.raw.raw_peaks AS
SELECT * EXCEPT (_rescued_data)
FROM read_files(
    '/Volumes/getdata/raw/source_files/peaks.csv',
    format => 'csv',
    header => true,
    inferSchema => true
);
```
