ნაწილი 2 : დოკუმენტირებული SELECT-ები

წაკითხვადი ვერსია აჩვენებს, რომელი სვეტი რას ნიშნავს.

```python
%sql
-- 2.1  მწვერვალების ცნობარი
SELECT
    Pk.peakid AS PeakId,-- უნიკალური კოდი (EVER).
    Pk.pkname AS PeakName,-- დასახელება.
    Pk.heightm AS HeightMetres,-- სიმაღლე. სკოუპი: 8000+.
    Pk.himal AS MountainRange,-- მთის ქედი.
    Pk.pstatus AS ClimbingStatus,-- Climbed / Unclimbed.
    Pk.pyear AS FirstAscentYear -- პირველი ასვლის წელი.
FROM getdata.raw.raw_peaks AS Pk
WHERE Pk.heightm >= 8000
ORDER BY Pk.heightm DESC;
```

```python
%sql
-- 2.2  ექსპედიციები
SELECT
    Ex.expid AS ExpeditionId, -- კოდი. უნიკალური არ არის (4 კოლიზია).
    Ex.peakid AS PeakId, -- FK raw_peaks-ზე.
    Ex.year AS ExpeditionYear, -- წელი. ჭეშმარიტების წყარო.
    Ex.season AS Season,-- სეზონი.
    Ex.agency AS Agency,-- ტრეკინგ-სააგენტო. გვერდი 3.
    Ex.comrte AS IsCommercialRoute, -- კომერციული ექსპედიცია.
    Ex.termreason AS TerminationReason, -- დასრულების მიზეზი.
    Ex.totmembers AS MemberCount, -- წევრები (დაქირავებულების გარეშე).
    Ex.tothired AS HiredStaffCount, -- დაქირავებული პერსონალი.
    Ex.o2used AS UsedOxygen -- დამატებითი ჟანგბადი.
FROM getdata.raw.raw_expeditions AS Ex
LIMIT 20;
```

```python
%sql
-- 2.3  მთამსვლელები
SELECT
    Mb.expid AS ExpeditionId,-- გარე გასაღები raw_expeditions-ზე.
    Mb.membid AS MemberNumber,-- ნომერი ექსპედიციის შიგნით.
    Mb.fname AS FirstName,-- სახელი.
    Mb.lname AS LastName,-- გვარი.
    Mb.sex AS Sex,-- M / F / X.
    Mb.yob AS BirthYear,-- დაბადების წელი (ასაკისთვის).
    Mb.citizen AS Citizenship,-- მოქალაქეობა.
    Mb.occupation AS Occupation,-- პროფესია.
    Mb.status AS ExpeditionRole,-- როლი (Climber, H-A Worker).
    Mb.hired AS IsHiredStaff,-- დაქირავებული თუ კლიენტი.
    Mb.msuccess AS ReachedSummit,-- მიაღწია თუ არა მწვერვალს.
    Mb.mo2used AS UsedOxygen, -- ჟანგბადი.
    Mb.death AS Died,-- დაიღუპა თუ არა.
    Mb.deathtype AS DeathType, -- მიზეზი. ცარიელი = ცოცხალია.
    Mb.deathclass AS DeathPhase -- ასვლის რომელ ეტაპზე.
FROM getdata.raw.raw_members AS Mb
LIMIT 20;
```

```python
%sql
--  2.4  14 რვაათასიანის სტატისტიკა (ვიკიპედია)
SELECT
    St.peak_name AS PeakName,-- მწვერვალის დასახელება.
    St.total_ascents AS TotalAscents,-- ასვლების რაოდენობა (ტექსტი).
    St.total_deaths AS TotalDeaths, -- დაღუპულთა რაოდენობა (ტექსტი).
    St.deaths_pct_of_ascents AS DeathsPctOfAscents, -- დაღუპულები ასვლების პროცენტად.
    St.climber_death_rate AS ClimberDeathRate  -- სიკვდილიანობა. en-dash = მონაცემი არ არსებობს.
FROM getdata.raw.raw_peak_stats_8000ers AS St
LIMIT 20;
```
