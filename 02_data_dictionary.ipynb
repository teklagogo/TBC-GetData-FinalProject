ნაწილი 2 : დოკუმენტირებული SELECT-ები

წაკითხვადი ვერსია აჩვენებს, რომელი სვეტი რას ნიშნავს.

```python
%sql
-- 2.1  მწვერვალების ცნობარი
SELECT
    pk.peak_id AS PeakId,-- მწვერვალის უნიკალური კოდი (PK).
    pk.peak_name AS PeakName, -- მწვერვალის ოფიციალური დასახელება.
    pk.peak_alternative_name AS PeakAlternateName, -- ალტერნატიული/ადგილობრივი დასახელება.
    pk.height_metres AS HeightMetres,-- სიმაღლე ზღვის დონიდან. ანალიზის მთავარი ცვლადი.
    pk.climbing_status AS ClimbingStatus,-- Climbed / Unclimbed.
    pk.first_ascent_year AS FirstAscentYear,  -- პირველი ასვლის წელი.
    pk.first_ascent_country AS FirstAscentCountry -- პირველი ამღები ქვეყანა.
FROM getdata.raw.raw_peaks AS Pk
LIMIT 20;
```

```python
%sql
-- 2.2  ექსპედიციები
SELECT
    ex.expedition_id AS ExpeditionId,-- ექსპედიციის უნიკალური კოდი.
    ex.peak_id AS PeakId,--FK raw_peaks-ზე.
    ex.year AS ExpeditionYear,-- ექსპედიციის წელი.
    ex.season AS Season,-- სეზონი (Spring/Summer/Autumn/Winter).
    ex.termination_reason AS TerminationReason, -- დასრულების მიზეზი. 'Success%' = მწვერვალი აღებულია.
    ex.highpoint_metres AS HighpointMetres,-- მიღწეული მაქსიმალური სიმაღლე.
    ex.members AS MemberCount, -- წევრების რაოდენობა (დაქირავებულების გარეშე).
    ex.member_deaths AS MemberDeaths,-- დაღუპული წევრები.
    ex.hired_staff AS HiredStaffCount,-- დაქირავებული პერსონალი (შერპები).
    ex.hired_staff_deaths AS HiredStaffDeaths,-- დაღუპული დაქირავებულები.
    ex.oxygen_used AS UsedOxygen -- დამატებითი ჟანგბადი: TRUE/FALSE.
FROM getdata.raw.raw_expeditions AS EX
LIMIT 20;
```

```python
%sql
-- 2.3  მთამსვლელები
SELECT
    mb.member_id AS MemberId,-- მონაწილის უნიკალური კოდი.
    mb.expedition_id AS ExpeditionId,-- FK raw_expeditions-ზე.
    mb.peak_id AS PeakId,-- FK raw_peaks-ზე.
    mb.citizenship AS Citizenship,-- მოქალაქეობა. ეროვნებების ანალიზის ველი.
    mb.sex AS Sex, -- სქესი: M / F.
    mb.age AS Age,-- ასაკი ექსპედიციის დროს (7-85).
    mb.expedition_role AS ExpeditionRole,-- როლი: Climber, H-A Worker, Leader...
    mb.hired AS IsHiredStaff,-- TRUE = დაქირავებული (შერპა).
    mb.success  AS ReachedSummit,-- მიაღწია თუ არა მწვერვალს.
    mb.oxygen_used AS UsedOxygen,-- გამოიყენა თუ არა ჟანგბადი.
    mb.died AS Died,-- დაიღუპა თუ არა. სიკვდილიანობის საფუძველი.
    mb.death_cause AS DeathCause,-- გარდაცვალების მიზეზი.
    mb.highpoint_metres AS HighpointMetres-- პირადი მაქსიმალური სიმაღლე.
FROM getdata.raw.raw_members AS Mb
LIMIT 20;
```
